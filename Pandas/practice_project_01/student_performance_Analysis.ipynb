### Create a new column score_level 
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
# Categorize each total score as High, Medium, or Low using the stated thresholds.
def score(x):
    if (x >= 90):
        return "High"
    elif(x >= 70 and x <= 90):
        return "Medium"
    elif(x < 70):
        return "Low"

```

```python
# Apply the score categorization function to each student's total score.
df['score_level'] = df['total_score'].apply(score)
```

### Create a new column study_category
"Light" if weekly_self_study_hours < 10, 
"Moderate" if 10 ≤ hours ≤ 20, 
"Heavy" if hours > 20

```python
# Categorize weekly study hours as Light, Moderate, or Heavy using the stated ranges.
def category(a):
    if a < 10:
        return "Light"
    elif a >= 10 and a <=20:
        return "Moderate"
    elif a > 20 :
        return "Heavy"
```

```python
# Apply the study-hour categorization function to every student.
df['study_category'] = df['weekly_self_study_hours'].apply(category)
```
