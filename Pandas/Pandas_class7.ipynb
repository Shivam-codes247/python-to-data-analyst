## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python
import pandas as pd

# Dataset 1: Product details
df1 = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Price": [55000, 800, 1500, 12000],
    "Stock": [10, 50, 30, 15],
    "Rating": [4.5, 4.2, 4.4, 4.6]
})

# Dataset 2: More products
# Same columns as df1 → good for vertical stacking
df2 = pd.DataFrame({
    "Product": ["Headphones", "Webcam", "Printer", "Speaker"],
    "Price": [2500, 3500, 8500, 3000],
    "Stock": [25, 18, 8, 20],
    "Rating": [4.3, 4.1, 4.0, 4.4]
})

# Dataset 3: Additional information
# Different columns → good for horizontal stacking
df3 = pd.DataFrame({
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics"],
    "Brand": ["Dell", "Logitech", "HP", "Samsung"],
    "Warranty_Years": [2, 1, 1, 3]
})

# Display them
df1
```

```python
pd.concat([df1,df2,df3], axis = 1) #for columns
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
import pandas as pd

sales_df = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Customer_ID": [1, 2, 3, 1, 4, 5, 2, 6],
    "Product_ID": [101, 102, 103, 101, 104, 105, 102, 106],
    "Quantity": [2, 1, 3, 1, 2, 4, 1, 2],
    "Sales": [110000, 800, 4500, 55000, 17000, 12000, 800, 6000]
})

sales_df
```

```python
df1.columns
```

```python
df1["Product_ID"] = [101, 102, 103, 104]
```

```python
df1
```

```python
pd.merge(df1,sales_df, on="Product_ID" )
```

```python
pd.merge(df1,sales_df, on="Product_ID", how = "right" )
```

```python
print("df1:", df1.columns.tolist())
print("sales_df:", sales_df.columns.tolist())
```

```python
pd.merge(df1,sales_df, on="Product_ID" , how = "left" )
```

```python
pd.merge(df1,sales_df, on="Product_ID" , how = "outer" )
```

```python
pd.merge(df1,sales_df, on="Product_ID" , how = "inner" )
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.
