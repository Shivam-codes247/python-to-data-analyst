## 1. Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd

data = {
    "Name": ["rahul kumar", "PRIYA SHARMA", "amit singh", "NEHA GUPTA", "rohan das", "ANJALI KUMARI"],
    "Age": [22, 24, 21, 25, 23, 26],
    "Maths": [85, 67, 92, 45, 78, 59],
    "Computer": [85, 67, 92, 45, 78, 59],
    "Salary": [25000, 32000, 28000, 40000, 30000, 35000],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"]
}

df = pd.DataFrame(data)

df
```

```python
#Convert all names to uppercase:

df['Name'] = df['Name'].map(str.upper) 

df
```

```python
df['Comp'] = [85,78,56,78,96,100]
```

```python
# df['Total_Marks'] = df[['Maths','Computer']].apply(
#     lambda row: row.sum() , axis=1
# )

df['Total_MRKS'] = df['Maths'] + df['Computer']

df
```

```python
col_mean = df[['Maths','Computer']].apply(
    lambda col : col.mean()
)

df

print(col_mean)
```

```python
def to_grade(s):
    if s>=90:
        return 'A'
    elif s>=80:
        return "B"
    elif s>=70:
        return 'C'
    else:
        return "Fail"


df['M_GRADE'] = df['Maths'].apply(to_grade)

df
```

```python
df = df.drop('MEAN',axis = 1)
```

```python
df['Name'] = df['Name'].map(str.lower)

df
```

```python
df['Maths_extras'] = df['Maths'].map(lambda x : x+5)

df
```
